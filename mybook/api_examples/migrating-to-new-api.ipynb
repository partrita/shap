# 새로운 "Explanation" API로 마이그레이션


```python
# An example dataset and model
import xgboost

import shap

X, y = shap.datasets.adult(n_points=100)
model = xgboost.XGBClassifier().fit(X, y)
explainer = shap.TreeExplainer(model, X)
```

API의 주요 변경 사항은 다음과 같습니다.


```python
shap_values = explainer.shap_values(X)  # Old style
explanation = explainer(X)  # New style
```

## SHAP 값 계산 방식의 변화

### 기존 방식 (Legacy Style)

`v0.36.0` 이전 버전에서는 SHAP 설명(Explanation)이 단순한 NumPy 배열로 반환되었으며, 주로 설명자(Explainer)의 `.shap_values()` 메서드를 통해 계산되었습니다.


```python
shap_values = explainer.shap_values(X)
shap_values[:2]  # a numpy array
```

마찬가지로 `shap.summary_plot`과 같은 레거시 플로팅 함수는 `shap_values`를 numpy 배열로 예상했습니다.

# 새로운 "Explanation" API로 마이그레이션


```python
explanation = explainer(X)
explanation[:2]  # a shap.Explanation object
```

`shap.Explanation` 객체는 단순히 SHAP 값(`.values` 속성)뿐만 아니라, 배경 데이터셋과 특성 이름(feature names) 같은 풍부한 맥락 정보를 포함하고 있습니다.

`shap.plots.bar`나 `shap.plots.beeswarm`과 같은 최신 시각화 함수들은 NumPy 배열 대신 이러한 `Explanation` 객체를 직접 입력으로 받아 더 상세한 시각화를 제공합니다.
